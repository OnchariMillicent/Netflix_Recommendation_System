## Outlier Detection and Treatment

Outlier detection is performed to identify numerical values that differ substantially from the overall distribution of the data. Such values may represent legitimate observations, data quality issues, or unusual characteristics of the Netflix content library.

The Interquartile Range (IQR) method is used to identify potential outliers by calculating the first quartile (Q1), third quartile (Q3), and IQR. Values below the lower boundary or above the upper boundary are flagged for further investigation.

The assessment focuses on relevant numerical features, including `release_year`, `date_added_year`, and `duration_value`. Movie runtimes and TV show season counts are evaluated separately because they represent different units of measurement. Calendar fields and `genre_count` are considered in context rather than automatically treated as continuous numerical variables.

Identified outliers are reviewed before any treatment is applied. Valid observations, such as older films or unusually long movies, are retained, while confirmed data quality issues are corrected where possible. Values are not automatically removed or capped solely because they fall outside the IQR boundaries.

This approach helps preserve meaningful information in the dataset while ensuring that any necessary data quality corrections are evidence-based.


```python
# Identify numerical columns and inspect their data types
netflix_df.select_dtypes(include="number").dtypes
```

```python
# Review summary statistics for all numerical columns
netflix_df.select_dtypes(include="number").describe().T
```

### **Checking potential outliers using the IQR method**

```python
# Define a reusable function to identify potential outliers using IQR
def check_outliers(data, column):
    # Calculate the first and third quartiles
    q1 = data[column].quantile(0.25)
    q3 = data[column].quantile(0.75)

    # Calculate the interquartile range
    iqr = q3 - q1

    # Define the lower and upper outlier boundaries
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    # Select values outside the boundaries
    outliers = data.loc[
        (data[column] < lower_bound) |
        (data[column] > upper_bound),
        column
    ]

    # Summarize the results
    return {
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier Values": sorted(outliers.unique().tolist())
    }
```

```python
# Assess potential outliers in original release years
check_outliers(netflix_df, "release_year")
```

```python
# Assess potential outliers in content addition years
check_outliers(netflix_df, "date_added_year")
```

```python
# Check movie runtimes in minutes
movie_durations_df = netflix_df.loc[netflix_df["type"] == "Movie"]
check_outliers(movie_durations_df, "duration_value")
```

```python
# Check TV Show season counts
tv_shows_df = netflix_df.loc[netflix_df["type"] == "TV Show"]
check_outliers(tv_shows_df, "duration_value")
```

#### Outlier Assessment observation

Outlier treatment was considered in the context of the Netflix dataset and the meaning of each numerical feature. Although statistical methods such as the Interquartile Range (IQR) can identify extreme values, these values do not necessarily indicate data quality issues.

The `release_year`, `date_added_year`, `date_added_month`, and `date_added_day` columns represent historical or calendar information. Unusually early release years may correspond to legitimate older titles, while month and day values follow calendar conventions. Therefore, these features are not subjected to automatic outlier removal or capping.

The `duration_value` feature is interpreted separately for Movies and TV Shows because it represents runtime in minutes for movies and season counts for television series. Unusually short or long durations are retained unless further investigation identifies an actual data quality issue. The `genre_count` feature is also retained because its values represent the number of genres associated with each title.

**Conclusion:** No automatic outlier removal or capping is applied. This approach preserves potentially meaningful observations while prioritizing the validity and context of the data over statistical thresholds alone.


### Correlation Assessment

Correlation analysis is used to examine the strength and direction of relationships between numerical variables. It can help identify patterns, potential redundancy, and relationships worth investigating further.

For this Netflix dataset, correlation analysis will be applied selectively because several numerical columns represent calendar components or engineered features rather than continuous measurements. The results will be interpreted in context, and correlation will not be treated as evidence of causation.


```python
# Select numerical features for an initial correlation assessment
correlation_features = netflix_df[
    ["release_year", "duration_value", "genre_count"]
]

# Calculate the Pearson correlation matrix
correlation_matrix = correlation_features.corr()

correlation_matrix
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Display correlations as a heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Matrix of Selected Netflix Features")
plt.tight_layout()
plt.show()
```

#### Correlation Analysis Findings

The initial Pearson correlation matrix shows a weak negative correlation between `release_year` and `duration_value` (\(r=-0.25\)). The relationship between `release_year` and `genre_count` is negligible (\(r=-0.04\)), while `duration_value` and `genre_count` also show a negligible positive correlation (\(r=0.09\)).

These results are preliminary because `duration_value` combines movie runtimes in minutes and TV show season counts. Consequently, correlations involving this feature may not represent meaningful relationships across the entire dataset. Further analysis should examine Movies and TV Shows separately before drawing conclusions about duration.


```python
# Correlations among selected numerical features for Movies
movie_correlation = netflix_df.loc[
    netflix_df["type"] == "Movie",
    ["release_year", "duration_value", "genre_count"]
].corr()

movie_correlation
```

```python
# Correlations among selected numerical features for TV Shows
tv_correlation = netflix_df.loc[
    netflix_df["type"] == "TV Show",
    ["release_year", "duration_value", "genre_count"]
].corr()

tv_correlation
```

#### Correlation Analysis by Content Type

To avoid mixing movie runtimes with TV show season counts, correlation analysis was performed separately for Movies and TV Shows.

For Movies, `duration_value` and `genre_count` show a moderate positive correlation (\(r=0.414\)), suggesting that movies associated with more genres tend to have longer runtimes. The correlation between `release_year` and `duration_value` is weakly negative (\(r=-0.206\)), while the relationship between `release_year` and `genre_count` is very weakly negative (\(r=-0.096\)).

For TV Shows, all three relationships are negligible. The correlation between `release_year` and `duration_value` is approximately -0.082, between `release_year` and `genre_count` is 0.087, and between `duration_value` and `genre_count` is -0.079.

Overall, the results indicate that the selected features have mostly weak linear relationships, with the association between movie duration and genre count standing out. These findings describe associations rather than causal relationships, and further analysis would be needed to investigate the underlying patterns.

### 5. Country VS Content Type

```python
# Select the content type and country columns
country_analysis = netflix_df[["type", "country"]].copy()

# Remove leading/trailing whitespace and convert missing values to empty strings
country_analysis["country"] = country_analysis["country"].fillna("").str.strip()

# Split multiple countries into individual entries
country_analysis["country"] = country_analysis["country"].str.split(", ")

# Create one row per country while retaining the associated content type
country_analysis = country_analysis.explode("country")

# Remove rows where the country information is missing or blank
country_analysis = country_analysis[
    country_analysis["country"].notna()
    & country_analysis["country"].ne("")
].reset_index(drop=True)

# Check the prepared data
country_analysis.head(10)
```

```python
# Count country entries for Movies and TV Shows
country_type_counts = pd.crosstab(
    index=country_analysis["country"],
    columns=country_analysis["type"]
)

# Identify the 10 countries with the most entries across both content types
top_countries = (
    country_type_counts.sum(axis=1)
    .sort_values(ascending=False)
    .head(10)
    .index
)

# Display the comparison table
country_type_counts.loc[top_countries].sort_values(
    by=["Movie", "TV Show"],
    ascending=False
)
```

```python
# Select the top 10 countries and sort them by total representation
country_chart_data = country_type_counts.loc[top_countries].copy()

country_chart_data["Total"] = country_chart_data.sum(axis=1)

country_chart_data = country_chart_data.sort_values("Total")

# Plot Movie and TV Show entries side by side
country_chart_data[["Movie", "TV Show"]].plot(
    kind="barh",
    figsize=(10, 6)
)

plt.xlabel("Number of Country–Title Associations")
plt.ylabel("Country")
plt.title("Top 10 Countries by Netflix Content Type")
plt.legend(title="Content Type")
plt.tight_layout()
plt.show()
```

#### Findings: Country vs. Content Type

The country analysis shows that the United States has the highest representation in both Movies and TV Shows, with 2,395 and 845 country-title associations, respectively. India has a substantially higher representation among Movies (976) than TV Shows (81), while the United Kingdom appears prominently in both categories.

Japan and South Korea have more TV Show associations than Movie associations in the top 10 countries examined. Pakistan also shows a notably higher representation among TV Shows, with 350 associations compared with 71 for Movies.

Overall, the results indicate differences in the country representation of Movies and TV Shows within the dataset. These patterns may help support country-based content exploration and recommendation features. However, titles can be associated with multiple countries, and the counts do not establish audience demand, subscriber preferences, or the total production output of each country.


## Exploratory Data Analysis — Summary and Key Insights

The Exploratory Data Analysis (EDA) examined the structure, composition, and characteristics of the Netflix content dataset to identify patterns relevant to content discovery and recommendation. The analysis covered content distribution, catalog additions, genre representation, ratings, release years, duration, feature relationships, and country representation across Movies and TV Shows.

### Key Findings

* **Content distribution:** Movies account for 6,126 titles, while TV Shows account for 2,664, making Movies the larger content category in the dataset.
* **Content additions:** The number of titles added to the dataset increased substantially during the late 2010s. More recent patterns may be influenced by the dataset's collection period and should not automatically be interpreted as changes in Netflix's overall catalog.
* **Genre representation:** International Movies, Dramas, and Comedies are among the most frequent Movie categories. International TV Shows, TV Dramas, and TV Comedies lead among TV Show categories. These patterns provide useful information for genre-based content exploration.
* **Content ratings:** TV-MA is the most common rating for both Movies and TV Shows, accounting for approximately 33.66% of Movies and 42.91% of TV Shows. The distributions also differ across other rating categories, including child-oriented ratings.
* **Release years:** TV Shows have a more recent release-year profile in this dataset, with a median release year of 2018 compared with 2016 for Movies. Movies span a broader range of release periods.
* **Content duration:** Movies have a median runtime of 98 minutes, while TV Shows have a median of one season. Duration was interpreted separately by content type because the numerical values represent different units.
* **Country representation:** The United States has the highest representation among both Movies and TV Shows in the country comparison. India has substantially more Movie than TV Show associations, while Japan, South Korea, and Pakistan have more TV Show than Movie associations within the top 10 countries examined. These counts represent country-title associations, not mutually exclusive production totals.
* **Feature relationships:** Correlation analysis identified a moderate positive relationship between movie duration and genre count (\(r=0.414\)). The relationship between movie release year and duration was weakly negative (\(r=-0.206\)). Most other relationships examined were weak or negligible, suggesting limited linear associations among the selected features.

### Implications for the Recommendation System

The findings support using content attributes such as genres, descriptions, cast, and directors to represent titles and identify similar content. Genre information can establish broad similarities between titles, while text-based features can capture additional information about their themes and subject matter. Country and content type may also support filtering or contextual recommendations where appropriate.

These attributes will inform the feature-preparation stage and help guide the development of a content-based recommendation system.

### Limitations and Considerations

The analysis describes the available dataset rather than Netflix's complete or current catalog. Genre and country associations can overlap because a title may have multiple genre labels or be associated with multiple countries. Correlations indicate associations rather than causal relationships, and catalog composition alone cannot establish subscriber preferences or viewing demand.

The dataset does not provide the user–item interaction data needed to directly evaluate personalized recommendations based on actual viewing behavior. Consequently, the initial recommendation system will focus on content similarity rather than claiming to predict individual preferences.

### Conclusion

The EDA established a foundation for feature preparation and content-based recommendation modeling. It identified important patterns in content type, genre, ratings, release periods, duration, and country representation while highlighting limitations that should guide subsequent modeling decisions.

The next stage will focus on inspecting and preparing recommendation-relevant features, developing content representations, and evaluating the quality and usefulness of the resulting recommendations.
